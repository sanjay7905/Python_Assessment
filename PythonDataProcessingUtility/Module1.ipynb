<h2>DataLoader</h2>

The DataLoader is used to read data from files.


```python
import os
import csv
import json

class DataLoader:

    def load(self,file_path):
        if not os.path.exists(file_path):
            print("File does not exist")
            return None

        extension = os.path.splitext(file_path)[1].lower()

        if extension == '.csv': 
            with open(file_path, 'r') as file:
                return list(csv.reader(file))
        elif extension == '.json':
            with open(file_path,'r') as file:
                return json.load(file)
        elif extension == '.txt':
            with open(file_path, 'r') as file:
                return file.read()
        else:
            print("File format is not supported")
            return None

    print("DataLoader Defined")
```

```python
loader = DataLoader()
data = loader.load('sample.txt')
print(data)
```
