<h2>Logging</h2>

The logging module is used to record information about what happens in our application.

```python
import logging
import os
```

<h3>logging module</h3>
It use to record messages such as:

Information
Warnings
Errors
Critical problems

<h3>os module</h3>

It use to work with folders and file paths.

<h2>Create logs folder</h2>
Before saving log messages, I create a folder named "logs".
os.makedirs() creates the folder.
exist_ok=True means Python will not give an error if the folder already exists.

```python
os.makedirs("logs", exist_ok=True)
```

<h2>Configure logging</h2>

logging.basicConfig() is used to configure how logging should work.
We specify where the log messages should be saved,
which level of messages should be recorded,
and how each log message should look.

```python
logging.basicConfig(
    filename="logs/application.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
```

```python
level=logging.INFO
```

INFO       - normal information
WARNING    - something may be wrong
ERROR      - an error occurred
CRITICAL   - serious problem

```python
format="%(asctime)s - %(levelname)s - %(message)s"
```

%(asctime)s - date and time
%(levelname)s - INFO, WARNING, ERROR.
%(message)s - our actual message

<h2>Create logger</h2>

getLogger() creates or gets a logger object.
The logger object is used to write messages into the log file.

```python
logger = logging.getLogger(__name__)
```

<h3>INFO message</h3>

```python
logger.info("Application started")
```

<h3>WARNING message</h3>

```python
logger.warning("This is a warning")
```

<h3>ERROR message</h3>

```python
logger.error("An error occurred")
```
